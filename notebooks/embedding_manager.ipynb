```python
from sentence_transformers import SentenceTransformer
import numpy as np
```

```python
class EmbeddingManager:
    
    def __init__(self, model_name:str= "all-MiniLM-L6-v2"):
        self.model_name=model_name
        self.model=None
        self._load_model()

    def _load_model(self):
        try:
            print(f"loading embedding model")
            self.model=SentenceTransformer(self.model_name)
            print(f"embeding model is loaded {self.model_name} with {self.model.get_embedding_dimension()}")
        except Exception as e:
            print(f"there is an exception {e}")
            raise
    
    def generate_embeddings(self, texts: list[str]) ->np.ndarray:

        if self.model is None:
            # print(f"there is no model it is not loaded {self.model_name}")
            raise ValueError(' model not loaded ')
        
        if len(texts) == 0:
            # print(f"texts are empty")
            raise ValueError('texts are empty')
        
        print(f" embeddings for the texts of size {len(texts)} ")
        embeddings=self.model.encode(texts, show_progress_bar=True, convert_to_numpy=True)
        print(f"embedding dimension of the texts is {embeddings.shape}")

        return embeddings
    
    

embedding_manager=EmbeddingManager()
embeddings=embedding_manager.generate_embeddings(['hey hi i am karthik','i am gay'])
```
